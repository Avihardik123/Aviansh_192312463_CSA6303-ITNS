```python
# Stateful Inspection Firewall Simulation

class StatefulFirewall:
    def __init__(self):
        self.connections = set()

    # Allow outbound connections and record their state
    def outbound(self, source_ip, source_port, dest_ip, dest_port):
        connection = (source_ip, source_port, dest_ip, dest_port)
        self.connections.add(connection)

        print("Outbound Connection:", connection)
        print("Status: ALLOWED")

    # Check incoming traffic
    def inbound(self, source_ip, source_port, dest_ip, dest_port):
        connection = (dest_ip, dest_port, source_ip, source_port)

        if connection in self.connections:
            print("Incoming Return Traffic: ALLOWED")
        else:
            print("Incoming Traffic: BLOCKED")


# Create firewall
firewall = StatefulFirewall()

# Test an outbound connection
firewall.outbound("192.168.1.10", 5000, "203.0.113.20", 443)

# Test return traffic for the connection
firewall.inbound("203.0.113.20", 443, "192.168.1.10", 5000)

# Test unsolicited inbound traffic
firewall.inbound("198.51.100.5", 22, "192.168.1.10", 6000)
```
