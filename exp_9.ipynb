```python
# Threat Intelligence Lifecycle Stage Classifier

def classify_activity(activity):
    text = activity.lower()

    rules = {
        "Direction": ["requirements", "objectives", "priority"],
        "Collection": ["collect", "gather", "logs", "feeds"],
        "Processing": ["normalize", "parse", "filter", "organize"],
        "Analysis": ["analyze", "correlate", "investigate", "impact"],
        "Dissemination": ["share", "publish", "distribute", "report"],
        "Feedback": ["feedback", "review", "improve", "lessons learned"]
    }

    for stage, keywords in rules.items():
        if any(word in text for word in keywords):
            return stage

    return "Stage not identified"


activity = input("Enter intelligence team activity: ")

stage = classify_activity(activity)

print("Threat Intelligence Lifecycle Stage:", stage)
```
