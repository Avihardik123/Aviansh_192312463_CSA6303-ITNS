```python
def classify_report(description, audience):
    text = (description + " " + audience).lower()

    strategic_keywords = [
        "business risk", "long-term", "executive",
        "board", "leadership", "investment"
    ]

    tactical_keywords = [
        "tactics", "techniques", "procedures",
        "ttp", "attack methods", "mitre"
    ]

    operational_keywords = [
        "ip address", "hash", "domain",
        "active incident", "ioc", "malware"
    ]

    strategic = sum(word in text for word in strategic_keywords)
    tactical = sum(word in text for word in tactical_keywords)
    operational = sum(word in text for word in operational_keywords)

    scores = {
        "Strategic": strategic,
        "Tactical": tactical,
        "Operational": operational
    }

    if max(scores.values()) == 0:
        return "Unclassified"

    return max(scores, key=scores.get)


description = input("Enter report description: ")
audience = input("Enter intended audience: ")

result = classify_report(description, audience)

print("Threat Intelligence Classification:", result)
```
