```python
import re
from collections import Counter

# Raw log data
raw_logs = [
    "2026-10-29 INFO user=avi action=login status=success",
    "2026-10-29 WARN user=ram action=login status=failed",
    "2026-10-29 WARN user=ram action=login status=failed",
    "2026-10-29 INFO user=avi action=logout status=success"
]

# Logstash: Parse raw logs
def parse_log(line):
    pattern = (
        r'(\S+) (\S+) user=(\S+) '
        r'action=(\S+) status=(\S+)'
    )

    match = re.match(pattern, line)

    if match:
        date, level, user, action, status = match.groups()

        return {
            "date": date,
            "level": level,
            "user": user,
            "action": action,
            "status": status
        }

    return None


events = [parse_log(line) for line in raw_logs]
events = [event for event in events if event]

# Elasticsearch: Store and index events
index = {
    i: event
    for i, event in enumerate(events, start=1)
}

print("Total indexed events:", len(index))

# Kibana: Aggregate and display results
level_counts = Counter(
    event["level"] for event in index.values()
)

status_counts = Counter(
    event["status"] for event in index.values()
)

print("\nEvents by log level:")
for level, count in level_counts.items():
    print(level, ":", count)

print("\nEvents by status:")
for status, count in status_counts.items():
    print(status, ":", count)

print("\nStructured log events:")
for event in index.values():
    print(event)
```
