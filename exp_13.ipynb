```python
def assess_threat_intelligence(data, context, confidence, actionable):

    if context and confidence >= 70 and actionable:
        return "Genuine Threat Intelligence"
    else:
        return "Unprocessed Raw Data"


data = input("Enter collected data: ")

context_input = input("Is context available? (yes/no): ").lower()
context = context_input == "yes"

confidence = int(input("Enter confidence percentage (0-100): "))

actionable_input = input("Is the information actionable? (yes/no): ").lower()
actionable = actionable_input == "yes"


result = assess_threat_intelligence(
    data,
    context,
    confidence,
    actionable
)

print("\nAssessment Result:", result)
```
