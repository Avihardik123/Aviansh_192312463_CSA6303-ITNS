```python
from urllib.parse import urlparse
import ipaddress

def check_url(url):
    if "://" not in url:
        url = "http://" + url

    parsed = urlparse(url)
    host = parsed.hostname or ""
    reasons = []

    # Check suspicious keywords
    keywords = ["login", "verify", "password", "account"]

    if any(word in url.lower() for word in keywords):
        reasons.append("Suspicious keyword found")

    # Check whether the domain is an IP address
    try:
        ipaddress.ip_address(host)
        reasons.append("IP address used as domain")
    except ValueError:
        pass

    # Check excessive subdomains
    if len(host.split(".")) > 4:
        reasons.append("Too many subdomains")

    # Check HTTPS
    if parsed.scheme != "https":
        reasons.append("HTTPS not used")

    print("URL:", url)

    if reasons:
        print("Result: Suspicious URL")
        for reason in reasons:
            print("-", reason)
    else:
        print("Result: No obvious phishing indicators")


url = input("Enter URL to check: ")
check_url(url)
```
