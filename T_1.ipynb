```python
def classify_scenario(scenario):
    text = scenario.lower()

    if any(word in text for word in
           ["bank", "payment", "transaction", "credit card"]):
        return "Banking Fraud Prevention"

    elif any(word in text for word in
             ["hospital", "patient", "medical", "healthcare"]):
        return "Healthcare Data Protection"

    elif any(word in text for word in
             ["online store", "e-commerce", "shopping", "marketplace"]):
        return "E-commerce Fraud Prevention"

    elif any(word in text for word in
             ["power grid", "water plant", "scada"]):
        return "Critical Infrastructure Protection"

    else:
        return "General Cyber Threat Intelligence"


scenario = input("Enter business scenario: ")

result = classify_scenario(scenario)

print("Threat Intelligence Application Area:", result)
```
