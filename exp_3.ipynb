```python
def check_policy(policy):
    checks = {
        "Scope": ["scope", "applies to", "coverage"],
        "Roles and Responsibilities": [
            "roles", "responsibilities", "owner"
        ],
        "Enforcement": [
            "enforcement", "disciplinary", "consequences"
        ],
        "Review Cycle": [
            "review cycle", "reviewed annually",
            "every year", "review date"
        ],
        "Approval Authority": [
            "approved by", "approval authority",
            "authorized by"
        ]
    }

    text = policy.lower()
    missing = []

    for element, keywords in checks.items():
        if any(word in text for word in keywords):
            print(element + ": FOUND")
        else:
            print(element + ": MISSING")
            missing.append(element)

    print("\nMissing Elements:")

    if missing:
        for element in missing:
            print("-", element)
    else:
        print("None")


policy = """
Scope: Applies to all company employees.
Roles and responsibilities are assigned to the security team.
The policy is reviewed annually.
"""

check_policy(policy)
```
