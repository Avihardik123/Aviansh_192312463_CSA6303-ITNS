```python
# Network Segmentation Simulation

# Define network access policies
POLICY = {
    ("User_Network", "Web_Server"): True,
    ("Web_Server", "Database"): True,
    ("Low_Trust", "Sensitive_Network"): False,
    ("Admin_Network", "Sensitive_Network"): True
}

# Function to check network access
def check_access(source, destination):
    if POLICY.get((source, destination), False):
        print(source, "->", destination, ": ACCESS ALLOWED")
    else:
        print(source, "->", destination, ": ACCESS BLOCKED")

# Test network connections
check_access("Low_Trust", "Sensitive_Network")
check_access("User_Network", "Web_Server")
check_access("Web_Server", "Database")
check_access("Unknown_Network", "Sensitive_Network")
```
